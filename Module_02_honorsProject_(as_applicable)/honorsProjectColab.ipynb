# What is the National Register?

The National Register of Historic Places (NRHP) is the official list of historic places in the United States that have been deemed worthy of preservation due to their significance in American history, architecture, archaeology, engineering, or culture. Established by the National Historic Preservation Act of 1966, the NRHP is managed by the National Park Service (NPS), which is a federal agency within the United States Department of the Interior.

The NRHP recognizes a wide range of historic properties, including buildings, structures, sites, districts, and objects. These properties can be of local, state, or national significance and are evaluated based on criteria such as their historical associations, architectural integrity, and contributions to American history and culture.

Listing in the National Register of Historic Places does not place any legal restrictions on what a private owner may do with their property unless federal funds or permits are involved. However, listing can provide recognition, access to certain preservation incentives, and protection against adverse effects from federally funded projects.

"On the evening of Friday, December 10, 2021, a potent storm system moving across the central United States resulted in widespread severe weather across the region, including a significant long track tornadoes. The code below visually represents the number of entries on the National Registry for each county specified in the counties_impacted list using a bar chart.







```python
import matplotlib.pyplot as plt

counties_impacted = ['Fulton', 'Hickman', 'Graves', 'Marshall', 'Lyon', 'Caldwell', 'Hopkins', 'Muhlenberg', 'Ohio', 'Breckenridge', 'Grayson']
# Creating the subset DataFrame
county_df = df[df['County'].isin(counties_impacted)]

# Count the number of entries for each county
county_counts = county_df['County'].value_counts()

# Plot the bar chart
fig, ax = plt.subplots(figsize=(10, 6))
county_counts.plot(kind='bar', ax=ax)
ax.set_title('Number of Entries on the National Registry by County')
ax.set_xlabel('County')
ax.set_ylabel('Number of Entries')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import geopandas as gpd
import zipfile
import os

# Read the KMZ file for counties
with zipfile.ZipFile('cnt_us.kmz', 'r') as zip_ref:
    zip_ref.extractall()
    kmz_files = [f for f in os.listdir('.') if f.endswith('.kmz')]
    for kmz_file in kmz_files:
        with zipfile.ZipFile(kmz_file, 'r') as zip_ref:
            zip_ref.extractall()
    extracted_files = [f for f in os.listdir('.') if not f.endswith('.kmz') and f.endswith(('.shp', '.geojson', '.kml'))]
    if extracted_files:
        counties = gpd.read_file(extracted_files[0])
    else:
        print("No supported geospatial file found in the extracted files.")

# List of tornado file paths
tornado_files = [
    '1210 Kentucky EF-4 Tornado.kmz',
    '1210 Muhlenberg County KY (Central City).kmz',
    '1210 TN-KY EF-3.kmz'
]

# Iterate over each tornado file
for i, tornado_file in enumerate(tornado_files, start=1):
    with zipfile.ZipFile(tornado_file, 'r') as zip_ref:
        zip_ref.extractall()
    extracted_files = [f for f in os.listdir('.') if not f.endswith('.kmz') and f.endswith(('.shp', '.geojson', '.kml'))]
    if extracted_files:
        tornado_path = gpd.read_file(extracted_files[-1])

        # Perform spatial join to identify counties intersected by the tornado path
        affected_counties = gpd.sjoin(counties, tornado_path, op='intersects')

        # Print the names of the affected counties for the current tornado
        print(f"Counties hit by Tornado {i}:")
        for county in affected_counties['county_name']:
            print(county)

        print()  # Add a blank line for separation between tornados
    else:
        print(f"No supported geospatial file found in the extracted files for Tornado {i}.")
```

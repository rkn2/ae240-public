## Non-Linear Models and Optimization

Exam 3 follows one design question from start to finish. You will clean a small dataset, compare groups, fit two non-linear models and pick the better one, and then use your model to find the best design. The practice exam walks through every one of these steps with a different design question, so work through it before the exam.

**Format.** Part 1 is take-home programming (open note). Part 2 is in class, handwritten and closed book. You will be given the outputs and plots from Part 1 and asked to explain them.

## I. Data Preparation and Comparing Groups

### A. Loading and cleaning
* Create a DataFrame from a dictionary I give you with `df = pd.DataFrame(data)`.
* Missing values are often written as text like `'NA'`. Replace them with `np.nan` and **save the result back** with `df = df.replace('NA', np.nan)`.
* Remove rows with missing values with `df = df.dropna()`, then print the DataFrame to check that the row is gone.
* On Colab you may see a pink `FutureWarning` about `replace`. That is fine, and your code still worked.

### B. Comparing groups with a loop
* Start with a list of the groups, like `types = ['A', 'B', 'C']`, and make two empty lists **before** the loop.
* Inside `for t in types:`, keep one group's rows with a boolean filter (`rows = df[df['column'] == t]`), then append `rows['value'].mean()` and `rows['value'].std()` to your lists.
* Make a bar chart with error bars with `ax.bar(types, means, yerr=stds, capsize=5)`, and give it a title and axis labels.
* Be able to explain what the **mean** tells you (the typical value for that group) and what the **standard deviation** tells you (how spread out the values are around it).
* Be able to use error bars to talk about **risk**. Long error bars mean the results are less predictable, so a group with a slightly higher mean but much longer error bars can be the riskier choice.

## II. Fitting Non-Linear Models

### A. Model functions and `curve_fit`
* Write a model as a function whose first input is `x` and whose other inputs are the parameters, like `def cubic_model(x, a, b, c, d):` that returns `a + b * x + c * x**2 + d * x**3`.
* Fit it with `popt, pcov = curve_fit(cubic_model, x_data, y_data)`. `popt` holds the fitted parameters in the same order as the function's inputs, so `popt[0]` is `a`, `popt[1]` is `b`, and so on. We don't use `pcov` in this class.
* Get the model's predicted values by calling the function with the fitted parameters, like `cubic_model(x_data, popt[0], popt[1], popt[2], popt[3])`.
* Know which shapes each model can make. A quadratic has one peak or valley and is symmetric around it. A cubic can rise quickly and fall slowly (or the other way around), so it can fit data that isn't symmetric.

### B. Comparing models with MSE
* Calculate the Mean Squared Error with `mean_squared_error(y_data, predicted_y)` from `sklearn.metrics`.
* MSE is the average of the squared misses between the data and the model. **Smaller is better**, so the model with the lower MSE fits the data better.

### C. Plotting the fits
* Make a scatter plot of the data, then plot each model as a smooth line over an x range like `x_range = pd.Series(range(0, 101))`.
* Every plot needs a title, x- and y-axis labels, and a legend when there is more than one line.

## III. Optimization

### A. `minimize` and finding a maximum
* `result = minimize(objective, starting_guess)` from `scipy.optimize` searches for the **lowest** value of `objective`. The best x is `result.x[0]` and the lowest value is `result.fun`.
* To find a **maximum**, minimize the **negative** of your model. Flipping the sign turns the peak into a valley that `minimize` can find, and `-result.fun` gives you the maximum value back.
* Write the objective as a function of `x` only, and use your fitted parameters inside it.
* `minimize` follows the slope downhill from the starting guess, so a bad starting guess can send it to the wrong place.

### B. The reliable recipe, grid search then `minimize`
* Use a `for` loop to build a list of x values (for example every whole number in your range), and calculate the model's value at each one.
* Put both lists in a DataFrame, find the best value with `.max()` (or `.min()`), and keep that row with a boolean filter, like `best_row = grid_df[grid_df['value'] == best_value]`.
* Start `minimize` from the best grid point. The answer should match what you found with a good starting guess.

### C. Limits of grid search
* A very fine grid is **slow**, it is only as **precise** as its spacing, and it only searches the **range** you choose.
* Be able to explain the **curse of dimensionality**. Every design variable multiplies the number of grid points (10 values for each of 3 variables is 1,000 points, and for 10 variables it is 10,000,000,000), so grid search becomes impractical for real designs with many variables. `minimize` only checks a handful of points.

## IV. Reading the Curve

* The **peak** (or the lowest point) is where the slope is **zero**.
* **Diminishing returns** means the slope is still positive but **getting smaller**, so each extra step of x helps less than the one before. Be able to spot this from the plot or from the change in y over equal steps of x.
* Near a peak the curve is almost flat, so a design a little below the peak gives up very little. Be able to explain why a designer might choose that design anyway (cost, energy, space, and so on).

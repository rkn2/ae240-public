## Variables

In Python, a variable is a symbolic name or identifier that represents a value or data stored in the computer's memory. Variables are used to store, manipulate, and retrieve data during the execution of a program. Think of a variable as a container that holds a value, and you can give it a meaningful name to represent what that value represents.

Key characteristics of variables in Python:

* **Assignment**: Variables are created and assigned values using the = operator. For example, `x = 10` assigns the value 10 to the variable `x`.

* **Dynamic Typing**: Python is dynamically typed, meaning you don't need to declare the data type of a variable explicitly. The interpreter determines the data type based on the assigned value. For example, `x = 10` assigns an integer to x, while `x = "Hello"` assigns a string.

* **Variable Names:** Variable names are case-sensitive and can consist of letters (a-z, A-Z), digits (0-9), and underscores (_). They must start with a letter or underscore, but *not* with a digit. For example, `my_variable`, `age`, and `_private_var` are valid variable names.

* **Reassignment:** You can change the value stored in a variable by assigning it a new value. For example, you can reassign `x` as follows: `x = 20`.

## Data Types

Now it turns out integers aren't the only type of data we'll interact with in Python.

Each data type in Python has its own set of methods and properties that can be used to manipulate and analyze the data. It's important to understand the different data types in order to effectively work with data in Python.

* **Integers**: whole numbers, positive or negative, without a fractional component. In Python, integers are represented by the `int` data type, like `diameter`.

* **Floats**: numbers with a decimal point, also known as floating-point numbers. In Python, floats are represented by the `float` data type, like `load`.

* **Strings**: sequences of characters, which can include letters, numbers, and symbols. In Python, strings are represented by the `str` data type, enclosed in quotes, like `material`.

* **Lists**: ordered collections of objects, which can be of different data types, enclosed in square brackets. We'll meet these in just a minute, once we load our full dataset.

* **Arrays**: similar to lists, but can only contain objects of the same data type. Used for numerical computations. We'll meet these in the Linear Regression lesson.

## Printing Readable Results

When you print a value with a label, the simplest approach is to separate them with **commas** in `print()`. Python adds spacing for you, and you don't need to convert numbers to strings.

````python
print("Radius:", radius, "inches")
print("Area:", area, "square inches")
print("Volume:", volume, "cubic inches")
````

### For the Curious: Other Ways to Print

You may see two other approaches online:
- **String concatenation** with `+`: join text and numbers, but you must wrap numbers in `str()` first.
- **f-strings**: prefix the quote with `f` and put variables in `{ }`. These also let you control formatting like decimal places.

Both work, but in this course we'll stick with the comma approach to keep things simple.

````python
# String concatenation (requires str() around numbers)
print("Radius: " + str(radius) + " inches")

# f-string (embed variables directly)
print(f"Radius: {radius} inches")

# f-string with formatting (3 decimal places)
print(f"Radius: {radius:.3f} inches")
````

## Lists

A **list** stores multiple values in order, enclosed in `[ ]`. We just turned `load` and `strain` into lists above, exactly what a list is for: holding many related values under one name.

````python
print(load)
print("Number of readings:", len(load))
````

## Lists Are Mutable: Adding and Removing Items

Lists can change after you create them. So far we've used functions like `print()` and `len()` where you hand the list TO the function. Lists also come with their own built-in actions that you call with a **dot** — think of it like the list knows how to do certain things to itself.

Think of these like wrenches in a toolbox. You don't need to know how the wrench was built — just which one to grab and when.

- `list_name.append(value)` adds `value` to the **end** of the list.
- `list_name.pop(index)` removes the item at `index` from the list.

### For the Curious: Why the Dot?
The dot syntax comes from Python's object system. Every list is an 'object' that carries its own set of actions (called 'methods'). When you write `load.append(1000)`, you're asking the `load` object to run its `append` method. If you want to explore further: [Python docs on list methods](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists).

````python
# Add an extra reading to the end of load
load.append(1000)
print(load)
print(len(load))
````

Now let's remove it again. Since we just added it to the end, we can use `.pop(-1)` to remove the last item:

````python
load.pop(-1)
print(load)
print(len(load))
````

`load` is back to the original 8 readings, exactly as we'll need it for everything that follows.

## Indexing

Access a specific item by its position (index). **Python counts from 0.**

```
load = [3945.5,  8548.7,  12494.2,  14993.1,  18807.1,  21042.9,  23410.3,  24857.0]
index:   0        1        2         3         4         5         6         7
```

Negative indices count from the end: `load[-1]` is the last item.

**Why start at 0?** The index is the *offset* from the start of the list. `load[0]` means "skip 0 items, give me the first one." `load[3]` means "skip 3, give me the next." This also makes slicing and `range()` math cleaner, as you'll see later.

````python
print(load[0])   # first value
print(load[-1])  # last value
print(load[3])   # fourth value
````

## Slicing

In the linear regression lesson, we'll need to analyze only the **elastic region** — the first 5 data points, where the material behaves predictably. We need a way to grab just *part* of a list.

Here's what that looks like in Excel — we selected just the linear portion of the data for the trendline:

Excel spreadsheet showing data selection and trendline

`list[start:stop]` extracts a sub-list. The `stop` index is **not included**.

Omit `start` to begin from the front; omit `stop` to go to the end.

````python
print(load[:4])   # first 4 values  (index 0, 1, 2, 3)
print(load[4:])   # last 4 values   (index 4, 5, 6, 7)
print(load[2:5])  # index 2, 3, 4
````

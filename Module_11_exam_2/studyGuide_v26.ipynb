Exam 2 follows one engineering job from start to finish. You will use a `while` loop to answer a design or scheduling question, clean a messy data report, label each row with `if`-`elif`-`else`, make plots, compare groups, and filter for the rows that need action. The practice exam walks through every one of these steps with a different job, so work through it before the exam.

**Format:** Part 1 is take-home programming (open note). Part 2 is in class and closed book: you will be given the outputs and plots from Part 1 and asked to explain them.

## I. Loops and Conditionals

### A. `while` loops
* A `while` loop repeats as long as its condition is `True`. Be able to write the condition from a sentence like "keep going until the target is reached" (e.g. `while total < target:`).
* Know **when** you need a `while` loop: when the number of steps is unknown ahead of time and each step depends on new input (like a daily reading), so there is no single formula to solve.
* Write every update out in full: `total = total + amount` and `day = day + 1`.
* Use a counter to step through a list inside the loop: `daily_values[day]` gets the value for the current day. Remember that Python counts from 0.
* Record each step by appending to lists before the loop starts and inside the loop, so you can plot the results afterward.
* Know what goes wrong, and trace the first few passes by hand to tell which one happens:
  * If nothing in the loop changes the condition, it never becomes `False` and you get an **infinite loop**.
  * If the counter never increases, the loop reuses the first value every pass. Whether that is an infinite loop **depends on that first value**: if it doesn't move the total toward the target (e.g. it is 0), the loop never ends; if it does, the loop ends but reports the wrong count.
  * If the list runs out before the condition becomes `False`, Python stops with an error (**`IndexError: list index out of range`**), because you asked for a list position that doesn't exist.
* Be able to explain how a change to the inputs (smaller or bigger daily values, a missing day) changes the result and why.

### B. `for` loops and `if`-`elif`-`else`
* Loop over a DataFrame column (`for value in df['column']:`) and append a label for each row to a list. Then add the list as a new column with `df['new_column'] = labels`.
* Use comparison operators: `>`, `<`, `==`, `!=`, `>=`, `<=`.
* Turn a set of thresholds into an `if`-`elif`-`else` chain, checking from one end of the range to the other.
* **`if`-`elif`-`else` vs. separate `if` statements:** `if`-`elif`-`else` stops at the first true condition, so each row gets exactly one label. With separate `if` statements, every true condition overwrites the one before it. Be able to trace a specific value through both versions and say which label it ends up with.

## II. Pandas Data Handling

### A. Loading and checking data
* Create a DataFrame from a dictionary I give you: `df = pd.DataFrame(data)`.
* Check column types with `df.dtypes`. Numbers stored as text show up as `object`.
* Count missing values with `df.isnull().sum()`.

### B. Cleaning missing values
* **Don't trust `.isnull().sum()` by itself.** Missing data is often written as placeholders like `'NA'`, `'-'`, or blanks. pandas reads those as ordinary text, so `isnull()` finds nothing.
* Replace placeholders with `np.nan` and **save the result back**: `df = df.replace(['NA', '-'], np.nan)`.
* Convert text columns to numbers with `.astype(float)`, then check `df.dtypes` again.
* Remove rows that are missing a value you need: `df = df.dropna(subset=['column_name']).copy()`. Know that rows removed this way are not included in any mean or standard deviation you calculate afterward.

### C. Filtering
* Filter rows with one condition: `df[df['column'] == value]` or `df[df['column'] > number]`.
* To find rows that meet **two** conditions, filter twice in a row and look at the result after each step.

### D. Comparing groups
* Count the rows in each category with `.value_counts()`.
* Compare groups with a loop:
  * Get the list of groups with `group_list = df['column'].unique()`.
  * Make two empty lists **before** the loop.
  * Inside `for group in group_list:`, keep one group with `one_df = df[df['column'] == group]`, then append `one_df['value'].mean()` and `one_df['value'].std()` to your lists.
* Standard deviation measures how spread out the values are. A bigger standard deviation means less consistent values.

## III. Plots and Interpretation

* Make each figure with `fig_name, ax_name = plt.subplots()` and show it by ending the cell with `fig_name`.
* **Line plot** of values you recorded in a loop: `ax.plot(x_list, y_list, marker='o', label='...')`. Add a target line with `ax.axhline(value, label='...')`, and show the labels with `ax.legend()`.
* **Bar chart** of category counts: `counts = df['column'].value_counts()`, then `ax.bar(counts.index, counts.values)`.
* **Histogram** of one numeric column with `ax.hist(df['column'], bins=5)`.
* **Bar chart with error bars:** `ax.bar(groups, means, yerr=stds, capsize=5)`. The **length** of each error bar shows that group's standard deviation; compare lengths, not how high the bars reach.
* Every plot needs a title and x- and y-axis labels.
* Be able to describe a histogram as roughly symmetric, **skewed right** (the long tail of bars stretches to the right), or **skewed left** (the tail stretches to the left), and say what that means for the data (for example, "most values are small and a few are very large").
* Be able to read a plot of a loop's results: why the line is steep in some places and flat in others, and where it crosses the target.
